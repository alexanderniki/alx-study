```python
# Читаем файлы

coderun: DataFrame = pd.read_csv(
    "data/data/itresume-coderun.csv",
    delimiter=",",
    header=0
)

users: DataFrame = pd.read_csv(
    "data/data/itresume-users-pandas.csv",
    delimiter=";",
    header=1
)
```

```python
# Смотрим, что там есть
coderun.head()
coderun.agg(["sum", "min"])
```

```python
# На всякий случай, посмотрим, порядок следования айдишников пользователей
coderun.sort_values(by="user_id", ascending=True)
```

```python
# Смотрим, что там есть
users.head()
```

```python
# На всякий случай, посмотрим, порядок следования айдишников пользователей
users.sort_values(by="id", ascending=True)
```

```python
# Мержим аналогично LEFT JOIN

df: DataFrame = coderun.merge(
    users, 
    left_on="user_id", 
    right_on="id", 
    how="left"
)
df.sort_values(by="user_id")
```

```python
df.rename(columns={
    "id_x": "id_run",
    "id_y": "id_user"
})
print(df)
```

```python
# сколько дней прошло между последним выполнением кода конкретного юзера и днем его регистрации. Столбец с разницей должен называться diff.

processed: DataFrame = df[["user_id", "created_at", "date_joined"]].sort_values(by="user_id", ascending=True)
processed
```

```python
# Проверим, что у нас есть реальные даты
processed[~processed["date_joined"].isna()].sort_values(by="user_id", ascending=True)
```

```python
# Приводим данные к типу datetime
processed["created_at"] = pd.to_datetime(processed["created_at"])
processed["date_joined"] = pd.to_datetime(processed["date_joined"])

# Считаем разницу и берем дни
processed["diff"] = (processed["created_at"] - processed["date_joined"]).dt.days
processed
```

```python
# Просто проверим, что у нас где-то посчиталась разница
processed[~processed["date_joined"].isna()].sort_values(by="diff", ascending=False)
```

```python
processed["diff"] = np.floor(processed["diff"] / 50) * 50
processed[~processed["date_joined"].isna()].sort_values(by="diff", ascending=False)
df2: DataFrame = processed
print(df2)
```

```python
df3: DataFrame = (
    df2.groupby("diff", dropna=False)["user_id"]
        .nunique()
        .sort_values(ascending=False)
        .reset_index(name="user_count")
)
print(df3)
```
